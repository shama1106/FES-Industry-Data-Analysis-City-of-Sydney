```python
import pandas as pd

file_path = r"C:\Users\91999\Downloads\FES_Industry_of_occupation.xlsx"
data = pd.read_excel(file_path, engine='openpyxl')

# Check for missing values in each column
missing_values_summary = data.isnull().sum()

print("Missing Values by Column:")
print(missing_values_summary)
```

```python
print(data.isnull().sum())  # Check for missing values
data_cleaned = data.dropna(subset=['X', 'Y'])
```

```python
# Check for missing values in the dataset
print("Missing Values by Column:")
print(data.isnull().sum())

# Drop rows with missing values in critical columns
data_cleaned = data.dropna(subset=['X', 'Y'])

# Verify the shape of the cleaned data
print(f"Data shape after dropping rows with missing values: {data_cleaned.shape}")
```

```python

duplicates = data_cleaned.duplicated(subset=['OBJECTID', 'ID', 'IndustryCode', 'Year', 'X', 'Y'])

# Drop duplicate rows
data_cleaned = data_cleaned.drop_duplicates(subset=['OBJECTID', 'ID', 'IndustryCode', 'Year' ,'X', 'Y'])

print(f"Data shape after removing duplicates: {data_cleaned.shape}")
```

```python
duplicates = data_cleaned.duplicated(subset=['X', 'Y','OBJECTID', 'ID', 'IndustryCode', 'Year'])
print(f"Number of duplicates: {duplicates.sum()}")
data_cleaned = data_cleaned.drop_duplicates(subset=['X', 'Y', 'OBJECTID', 'ID', 'IndustryCode', 'Year'])
```

```python
data_cleaned['DivisionName'] = data_cleaned['DivisionName'].str.lower().str.strip()
data_cleaned['IndustryName'] = data_cleaned['IndustryName'].str.lower().str.strip()
data_cleaned['Village'] = data_cleaned['Village'].str.lower().str.strip()
```

```python
print(data_cleaned[['X', 'Y']].describe())  # Check ranges
data_cleaned = data_cleaned[(data_cleaned['X'] > 333000) & (data_cleaned['X'] < 334000) &
                            (data_cleaned['Y'] > 6250000) & (data_cleaned['Y'] < 6260000)]
```

```python
data_cleaned['Location Summary'] = data_cleaned['Village'] + " - " + data_cleaned['DivisionName']
```

```python
# Check the list of columns to ensure 'Location Summary' was added
print(data_cleaned.columns)
```
